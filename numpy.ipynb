```python
marks = np.array([70,80,90,99,67,86])
print(marks)
print(type(marks))

print("Reshape Array:")
print(marks.reshape(2,3))

print("min value:",marks.min())
print("max value:",marks.max())
print("mean of marks:",marks.mean())

print("marks greater than 40 and less than 80:")
print(marks[(marks>40) &(marks<80)])
```

```python
mark=np.array([55,72,88,41,95,63])
names = np.array(["aisha","ben","carla","dev","eli","farah"])
print("name:",names)
print("marks:",mark)
print("name where everyone who scored above 70:",names[mark>70])

count = np.sum((mark>50) & (mark<80))
print("how many people scored between 50 and 80:",count)

print(np.where(mark>=80,"A","B"))
```

```python
import numpy as np
# ---- THE DATA ----

student_names = np.array(["Aisha", "Ben", "Carla", "Dev", "Eli", "Farah"])   # six students
subject_names = np.array(["Maths", "Physics", "English"])                    # three subjects

scores = np.array([[78, 65, 82],    # Aisha's three marks - one per subject, in the order above
                   [55, 71, 60],    # Ben's
                   [91, 88, 79],    # Carla's
                   [42, 38, 55],    # Dev's
                   [67, 74, 71],    # Eli's
                   [83, 97, 68]])   # Farah's

print(scores)                       # the whole table
print("shape:", scores.shape)

for i in range(len(student_names)):
  name = student_names[i]
  total =  scores[i].sum()
  avg = scores[i].mean()
  print(name ,"total :",total ,"avg:",avg)

avg_subject = scores.sum(axis=0)
print(avg_subject)
print("Hardest subject:",subject_names[avg_subject.argmin()])
print("Easiest subject:",subject_names[avg_subject.argmax()])
```
